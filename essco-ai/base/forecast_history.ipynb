```python
import requests
import pandas as pd

# Define parameters for your exact location
params = {
    "latitude": 14.066094225657675,
    "longitude": 100.63597818535003,
    "start_date": "2026-05-11",  # 30 days ago
    "end_date": "2026-06-11",    # Today
    "hourly": "temperature_2m,relative_humidity_2m,cloud_cover,shortwave_radiation",
    "timezone": "Asia/Bangkok"
}

response = requests.get("https://historical-forecast-api.open-meteo.com/v1/forecast", params=params).json()

# Convert the hourly array directly into a Pandas DataFrame
df_forecast = pd.DataFrame(response["hourly"])
df_forecast['time'] = pd.to_datetime(df_forecast['time'])
df_forecast.set_index('time', inplace=True)

# Rename columns to match your exact feature list
df_forecast.columns = [
    'ambient_temp_forecast_1h', 
    'humidity_forecast_1h', 
    'cloud_cover_forecast_1h', 
    'solar_irradiance_forecast_1h'
]

df_forecast.head(24)
```

```python
# Resample to 5-minute grid and fill the hourly gaps smoothly
df_forecast = df_forecast.resample('5min').mean().interpolate(method='linear')
df_forecast.head(100)
```
